```python
"""
[YouTube - Steve Brunton - Linear Regression 2](https://www.youtube.com/watch?v=88_wgLgx0_k&list=PLMrJAkhIeNNSVjnsviglFoY2nXildDCcv&index=20)
"""
import numpy as np
import matplotlib.pyplot as plt

A = np.loadtxt("hald_ingredients.csv", delimiter=",")
b = np.loadtxt("hald_heat.csv", delimiter=",")

U, S, VT = np.linalg.svd(A, full_matrices=False)
x = VT.T @ np.linalg.inv(np.diag(S)) @ U.T @ b

plt.figure(figsize=(8, 8))
plt.plot(b, color="k", label="Raw Data")
plt.plot(A@x, "-o", color="r", label="Regression")
plt.legend()
plt.show()
```

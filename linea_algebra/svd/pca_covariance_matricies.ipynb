# 📘 **Notes on SVD, PCA, and Covariance Matrices**

## **1. Data Matrix Setup**
Let $A$ be a data matrix with:
- **Rows = samples**
- **Columns = features**

Before doing PCA or interpreting covariance, the data must be **mean‑centered** (subtract the mean of each column).

---

## **2. Covariance Matrix of the Features**
The sample covariance matrix of the features is:

$$
\Sigma = \frac{1}{n-1} A^T A
$$

This matrix describes how **features vary together** across samples.  
It is the matrix PCA diagonalizes.

- Large diagonal entries → high variance in a feature  
- Off‑diagonal entries → correlation between features

---

## **3. Relationship to SVD**
The Singular Value Decomposition of $A$ is:

$$
A = U \Sigma V^T
$$

From this, two important matrices appear:

### **(a) $A^T A$**  
$$
A^T A = V \Sigma^2 V^T
$$

- Describes relationships between **columns/features**  
- Eigenvectors = principal directions in feature space  
- Used directly in PCA

### **(b) $A A^T$**  
$$
A A^T = U \Sigma^2 U^T
$$

- Describes relationships between **rows/samples**  
- Sometimes called the **Gram matrix** of the samples  
- Useful when the number of samples is small compared to the number of features

---

## **4. Important Distinction**
Although $A^T A$ and $A A^T$ resemble covariance matrices, they are **not** covariance matrices unless:

1. The data is **mean‑centered**, and  
2. You divide by $n-1$.

Only then does:

$$
\frac{1}{n-1} A^T A
$$

become the true covariance matrix of the features.

---

## **5. Mental Model**
- **$A^T A$** → similarity of **features**  
- **$A A^T$** → similarity of **samples**  
- **Covariance matrix** → centered and scaled version of $A^T A$

If you use **$n$** in the denominator when you *should* have used **$n-1$**, you end up with a **biased underestimate of the true variance**. That’s the whole reason Bessel’s correction exists.

Here’s the clean breakdown.

---

## 🎯 What goes wrong when you divide by $n$ for a sample?

### **1. The variance becomes systematically too small**
When you compute sample variance, you use the **sample mean**, not the true population mean.  
The sample mean is *closer* to the sample data than the true mean would be, so the squared deviations shrink.

If you divide by $n$, you don’t compensate for this shrinkage.

Result:

$$
\text{Var}_{\text{biased}} = \frac{1}{n} \sum (x_i - \bar{x})^2
$$

This is **biased downward**.

---

## ⭐ Why does dividing by $n-1$ fix it?

Because using the sample mean removes one degree of freedom.

- You have $n$ data points.
- But once you compute the sample mean, only $n-1$ of the deviations can vary freely.
- The last deviation is forced to make the average zero.

So dividing by $n-1$ corrects the bias:

$$
\text{Var}_{\text{unbiased}} = \frac{1}{n-1} \sum (x_i - \bar{x})^2
$$

This gives an **unbiased estimator** of the population variance.

---

## 📉 How big is the error?

The biased estimator is smaller by a factor of:

$$
\frac{n-1}{n}
$$

So the underestimate is:

$$
\text{Underestimate} = \frac{1}{n}
$$

For example:

- If $n = 10$, you underestimate variance by **10%**.
- If $n = 100$, you underestimate variance by **1%**.
- If $n = 1{,}000$, the difference is tiny.

This is why large datasets often don’t care which denominator you use — but small datasets absolutely do.

---

## 🧠 Intuition
Using $n$ pretends you already know the true mean.  
But you don’t — you estimated it from the data.

Estimating the mean “uses up” one degree of freedom, so you must divide by $n-1$ to compensate.

---

## 🎯 Takeaway
- **Using $n$ for a sample → biased, too small variance.**
- **Using $n-1$ → unbiased estimate of population variance.**
- **Using $n$ is correct only when you truly have the entire population.**

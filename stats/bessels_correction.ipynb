If you use **$n$** in the denominator when you *should* have used **$n-1$**, you end up with a **biased underestimate of the true variance**. That’s the whole reason Bessel’s correction exists.

Here’s the clean breakdown.

---

## 🎯 What goes wrong when you divide by $n$ for a sample?

### **1. The variance becomes systematically too small**
When you compute sample variance, you use the **sample mean**, not the true population mean.  
The sample mean is *closer* to the sample data than the true mean would be, so the squared deviations shrink.

If you divide by $n$, you don’t compensate for this shrinkage.

Result:

$$
\text{Var}_{\text{biased}} = \frac{1}{n} \sum (x_i - \bar{x})^2
$$

This is **biased downward**.

---

## ⭐ Why does dividing by $n-1$ fix it?

Because using the sample mean removes one degree of freedom.

- You have $n$ data points.
- But once you compute the sample mean, only $n-1$ of the deviations can vary freely.
- The last deviation is forced to make the average zero.

So dividing by $n-1$ corrects the bias:

$$
\text{Var}_{\text{unbiased}} = \frac{1}{n-1} \sum (x_i - \bar{x})^2
$$

This gives an **unbiased estimator** of the population variance.

---

## 📉 How big is the error?

The biased estimator is smaller by a factor of:

$$
\frac{n-1}{n}
$$

So the underestimate is:

$$
\text{Underestimate} = \frac{1}{n}
$$

For example:

- If $n = 10$, you underestimate variance by **10%**.
- If $n = 100$, you underestimate variance by **1%**.
- If $n = 1{,}000$, the difference is tiny.

This is why large datasets often don’t care which denominator you use — but small datasets absolutely do.

---

## 🧠 Intuition
Using $n$ pretends you already know the true mean.  
But you don’t — you estimated it from the data.

Estimating the mean “uses up” one degree of freedom, so you must divide by $n-1$ to compensate.

---

## 🎯 Takeaway
- **Using $n$ for a sample → biased, too small variance.**
- **Using $n-1$ → unbiased estimate of population variance.**
- **Using $n$ is correct only when you truly have the entire population.**

# 🎯 **Why the last deviation is forced to make the average zero**

Suppose you have a sample of $n$ numbers:

$$
x_1, x_2, \dots, x_n
$$

You compute the **sample mean**:

$$
\bar{x} = \frac{1}{n} \sum_{i=1}^n x_i
$$

Now consider the deviations from the mean:

$$
d_i = x_i - \bar{x}
$$

These deviations always satisfy a strict identity:

$$
\sum_{i=1}^n d_i = 0
$$

This is not a coincidence — it is *forced* by how the mean is defined.

---

# 🧠 **What this means for degrees of freedom**

Let’s say you choose the first $n-1$ deviations freely:

$$
d_1, d_2, \dots, d_{n-1}
$$

The last deviation **cannot** be chosen freely. It must satisfy:

$$
d_n = -\sum_{i=1}^{n-1} d_i
$$

If it didn’t, the deviations wouldn’t sum to zero, and they wouldn’t correspond to any real sample mean.

So:

- You have **n deviations**
- But only **n−1 can vary independently**
- The last one is **determined** by the others

This is the essence of **losing one degree of freedom**.

---

# 📉 **How this affects variance**

Variance uses squared deviations:

$$
\sum_{i=1}^n (x_i - \bar{x})^2
$$

But because the deviations are constrained, this sum is **systematically smaller** than it would be if you knew the true population mean.

Dividing by $n$ assumes all deviations are independent — but they aren’t.

Dividing by $n-1$ corrects for the fact that one deviation is “locked in.”

---

# 🔍 **A tiny numerical example**

Take three numbers:

$$
x_1 = 5,\quad x_2 = 7,\quad x_3 = 9
$$

Mean:

$$
\bar{x} = 7
$$

Deviations:

$$
d_1 = -2,\quad d_2 = 0,\quad d_3 = 2
$$

Notice:

$$
-2 + 0 + 2 = 0
$$

Now imagine you try to “choose” deviations freely:

If you pick:

$$
d_1 = -2,\quad d_2 = 0
$$

Then $d_3$ **must** be:

$$
d_3 = -(-2 + 0) = 2
$$

You cannot choose it independently.

---

# 🎯 **Takeaway**

- When you compute the sample mean, you impose a constraint:  
  **the deviations must sum to zero.**
- This removes one degree of freedom.
- Therefore, the unbiased variance estimator divides by **n−1**, not **n**.
- Using $n$ underestimates variance because it pretends all deviations are independent.
